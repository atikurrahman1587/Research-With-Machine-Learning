```python
choice = int(input("1. Celsius to Fahrenheit\n2. Fahrenheit to Celsius\n"))

if choice == 1:
    c = float(input("Enter Celsius: "))
    f = (c * 9/5) + 32
    print("Fahrenheit:", round(f, 2))
else:
    f = float(input("Enter Fahrenheit: "))
    c = (f - 32) * 5/9
    print("Celsius:", round(c, 2))
```
